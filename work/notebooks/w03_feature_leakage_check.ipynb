## 1. Build the feature vector

Code that actually builds it — engineered features, categorical handling, fills. This cell builds a reproducible feature matrix `X` and label `y` from the starter CSV `data/raw/content_refresh_anonymized.csv`.

Notes of caution applied from `skills/flyrank/flyrank-data/SKILL.md`: rate columns are percentages scaled ×100 (we convert to fractions), `avg_position == 0` means missing (we create a has_position flag), and `trend_pct`/`trend_direction` are derived from the label and must not be used in honest features.


```python
# Build feature vector
import pandas as pd
import numpy as np
from pathlib import Path

SEED = 42
np.random.seed(SEED)

DATA_PATH = Path('data/raw/content_refresh_anonymized.csv')
assert DATA_PATH.exists(), f'Expected starter CSV at {DATA_PATH} — make sure you ran setup or downloaded the repo data.'
df = pd.read_csv(DATA_PATH)

# Quick probe (keeps output small when run):
print('rows,cols:', df.shape)
print('base rate (label) name candidate columns:', [c for c in df.columns if 'label' in c or 'trend' in c])

# Define label from repo docs (starter dataset uses `is_declining_label`)
assert 'is_declining_label' in df.columns, 'Notebook expects is_declining_label to exist in the starter CSV'
y = df['is_declining_label'].astype(int)

# Start feature engineering: safe, explainable features only
X = pd.DataFrame(index=df.index)
# IDs for grouping / splitting only (not features)
X['client_id'] = df['client_id']
X['content_id'] = df['content_id']

# content_type: low-cardinality categorical -> one-hot (drop_first to avoid collinearity)if 'content_type' in df.columns:
    ct = pd.get_dummies(df['content_type'].fillna('MISSING'), prefix='content_type', dtype=int)
    # keep columns but don't include the full set in tree models if you prefer; drop_first keeps parity
    ct = ct.loc[:, ct.columns.sort_values()]
    X = pd.concat([X, ct], axis=1)

# word_count: numeric, has missingness correlated with content_type (docs warn)
if 'word_count' in df.columns:
    X['word_count'] = df['word_count'].replace(0, np.nan)
    X['has_word_count'] = (~X['word_count'].isna()).astype(int)
    X['word_count'] = X['word_count'].fillna(df['word_count'].median())

# avg_position: 0 means no data — create flag and impute to median of real positions
if 'avg_position' in df.columns:
    pos = df['avg_position'].replace(0, np.nan)
    X['has_position'] = (~pos.isna()).astype(int)
    X['avg_position'] = pos.fillna(pos.median())

# Rate columns: per SKILL.md they are ×100 percentages (eg 0.76 == 0.76%). Convert to fractions in [0,1].
rate_cols = [c for c in df.columns if c in ('ctr','engagement_rate','scroll_rate','ai_traffic_pct','trend_pct')]
# We'll include safe rate columns except trend_pct (it's label-derived) — see exclusions below
for c in ['ctr','engagement_rate','scroll_rate','ai_traffic_pct']:
    if c in df.columns:
        X[c] = pd.to_numeric(df[c], errors='coerce') * 0.01

# IMPORTANT: trend_pct and trend_direction are listed in the repo SKILL as label-derived. Do NOT include them in X for honest training.
# Keep a separate copy for leakage-testing below (we do not use them in honest features):
leak_cols = {}
for c in ['trend_pct','trend_direction']:
    if c in df.columns:
        leak_cols[c] = df[c]

# Fill any remaining numeric NaNs with medians (conservative)
for c in X.select_dtypes(include=[np.number]).columns:
    X[c] = X[c].fillna(X[c].median())

# Drop pure-ID columns from model features (client_id/content_id) but keep for grouped splitting — move them out before modeling
group_cols = X[['client_id','content_id']].copy()
X_model = X.drop(['client_id','content_id'], axis=1)

print('Feature matrix ready — shape', X_model.shape)
# small save so the rest of the notebook can inspect without rerunning the feature build
Path('work/outputs').mkdir(parents=True, exist_ok=True)
X_model.to_parquet('work/outputs/w03_features.parquet')
pd.Series(y).to_frame('is_declining_label').to_parquet('work/outputs/w03_label.parquet')

# Expose X, y, and group_cols for following cells to use when the notebook is run top-to-bottom
X = X_model
# restore y as Series
y = pd.Series(y, name='is_declining_label')
groups = group_cols['client_id']

print('Base rate (positive fraction):', y.mean())
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

For each feature included above I record meaning, missingness handling, and whether the feature is available at prediction time. Keep this human text in the markdown cell and the code above as the provenance.

- content_type_* : one-hot of content type. Missing -> 'MISSING'. Available at prediction time and safe.
- word_count : count of words. Missingness handled with has_word_count flag and median imputation; available at prediction time.
- has_word_count : boolean flag that encodes whether word_count was present (missingness correlates with content_type).
- has_position : 1 if avg_position was available (avg_position==0 means no data). Separately use avg_position imputed; the flag prevents hiding missingness signal.
- avg_position : median-imputed average rank (0 in raw means no data; we treat 0 as missing). Available at prediction time only if position data exists.
- ctr, engagement_rate, scroll_rate, ai_traffic_pct : converted from ×100 percentages to fractions (divide by 100). For example, raw 0.76 -> 0.0076. Missing coerced to numeric and median-imputed. These are available at prediction time per the starter CSV trailing-window semantics.

Note: I deliberately exclude `trend_pct` and `trend_direction` from the honest feature set because they are label-derived (the label `is_declining_label` is computed from them). See the exclusions list below for full rationale.

## 3. The leakage hunt

Attack your own features: we run grouped cross-validation (GroupKFold by client_id) and report base rate + honest model score. Then deliberately add the forbidden `trend_pct` column to demonstrate leakage: if the score jumps dramatically and `trend_pct` becomes the top feature, that's a confession.


```python
# Leakage tests: grouped CV + leaky-feature experiment
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.metrics import roc_auc_score, accuracy_score
import numpy as np

SEED = 42
clf = RandomForestClassifier(n_estimators=200, random_state=SEED, n_jobs=-1)
gkf = GroupKFold(n_splits=5)

print('Base rate (positive fraction):', y.mean())
# Honest score (no label-derived features)
scores = cross_val_score(clf, X, y, groups=groups, cv=gkf, scoring='roc_auc', n_jobs=-1)
print('Honest AUC (GroupKFold):', np.round(scores.mean(), 4), '+/-', np.round(scores.std(),4))

# Now add a leaky feature (trend_pct) to demonstrate leakage — DO NOT USE in honest production training
if 'trend_pct' in leak_cols:
    X_leaky = X.copy()
    # trend_pct in the starter CSV is ×100 percentage; convert the same way we converted ctr etc.
    X_leaky['trend_pct_leak'] = pd.to_numeric(leak_cols['trend_pct'], errors='coerce') * 0.01
    scores_leak = cross_val_score(clf, X_leaky, y, groups=groups, cv=gkf, scoring='roc_auc', n_jobs=-1)
    print('Leaky AUC with trend_pct included (GroupKFold):', np.round(scores_leak.mean(),4), '+/-', np.round(scores_leak.std(),4))
    # Train one model to show feature importances including the leaky feature
    clf.fit(X_leaky, y)
    importances = pd.Series(clf.feature_importances_, index=X_leaky.columns).sort_values(ascending=False)
    print('
Top 10 feature importances (leaky model):')
    print(importances.head(10))
else:
    print('trend_pct not present in the starter CSV — leakage test cannot add trend_pct, but follow the same pattern if you run against the full warehouse.')
```

## 4. What I excluded and why

- trend_pct : label-derived (used to compute `trend_direction` and the provided label). Including it leaks the answer — excluded from honest features.
- trend_direction : derived from `trend_pct` and directly used to create `is_declining_label` — excluded.
- content_id, client_id : pseudonymous IDs — useful only for grouping and splits; if used directly as features they leak client-specific signals and enable memorization.
- any product/decision flags (none present in the starter CSV) : these reflect past decisions and produce circular learning.
